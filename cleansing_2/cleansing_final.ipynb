```python
import pandas as pd

# Bersihkan tanda kutip pada kolom "alamat sppg" di file cleaned_cleanser.csv

file_path = "cleaned_cleanser1.csv"

# Baca file CSV
df = pd.read_csv(file_path)

# Pastikan kolom ada
if "Alamat SPPG" in df.columns:
    # Hapus semua tanda kutip (" dan ') dari isi kolom
    df["Alamat SPPG"] = (
        df["Alamat SPPG"]
        .astype(str)
        .str.replace('"', "", regex=False)
        .str.replace("'", "", regex=False)
        .str.strip()
    )

    # Simpan hasilnya kembali ke file yang sama
    df.to_csv(file_path, index=False, encoding="utf-8")
    print("Kolom 'alamat sppg' sudah dibersihkan dari tanda kutip.")
else:
    print("Kolom 'alamat sppg' tidak ditemukan di file CSV.")
```

```python
# Cek apakah masih ada tanda kutip (" atau ') di file CSV
# file_path sudah ada dari cell sebelumnya: "cleaned_cleanser1.csv"

df_check = pd.read_csv(file_path)

cols_with_quotes = []

for col in df_check.select_dtypes(include=["object"]).columns:
    mask = df_check[col].astype(str).str.contains(r'["\']', regex=True, na=False)
    count = int(mask.sum())
    if count > 0:
        cols_with_quotes.append((col, count))

if cols_with_quotes:
    print("Masih ada tanda kutip pada kolom berikut:")
    for col, count in cols_with_quotes:
        print(f" - {col}: {count} baris")
else:
    print("Tidak ada tanda kutip pada semua kolom string di file CSV.")
```
