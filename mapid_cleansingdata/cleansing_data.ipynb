```python
import pandas as pd
```

```python
input_file = 'split_cleanser1.csv'
output_file = 'data_cleansing_1.csv'

column = ['title','address','latitude','longitude','link']

df = pd.read_csv(input_file, encoding='utf-8-sig')

print(f"baris awal: {len(df):,}")

df = df[[c for c in column if c in df.columns]].copy()
print(df.head())

# print(df.columns)
# print(df['title'])
```

```python
df = df.rename(columns={
    'title':'Nama SPPG',
    'address':'Alamat SPPG',
    'latitude':'Latitude',
    'longitude':'Longitude',
    'link':'link',
})


df.insert(0, 'No SPPG', range(1, len(df)+1)) 

for col in ['Provinsi SPPG','Kabupaten/Kota SPPG','Kecamatan SPPG']: 
    if col not in df.columns: 
        df[col] = '' 

 

final_column = ['No SPPG','Nama SPPG','Provinsi SPPG', 

               'Kabupaten/Kota SPPG','Kecamatan SPPG', 

               'Latitude','Longitude','Alamat SPPG','link'] 

df = df[final_column] 
```

```python
print(df.isnull())
print(df.isnull().sum())

print(df.isnull().values.any())
```

```python
null = df[df.isnull().any(axis=1)]
print(null)
```

```python
import re

def cleaning_address(text):
    if pd.isna(text):
        return ""

    text = str(text) 
    text = re.sub(r'\b[A-Z0-9]{4}\+[A-Z0-9]{2,3}\b', '', text)  # hapus plus code 
    text = re.sub(r'[#@\^\$\?]', '', text)                        # hapus simbol 
    text = re.sub(r'^[,\s]+', '', text)                             # hapus awal koma 
    text = text.sub(' ,', ',')                                   # spasi sebelum koma 
    text = re.sub(r'\s+', ' ', text).strip() 
    return text 

df['Alamat SPPG'] = df['Alamat SPPG'].apply(cleaning_address)
df['Nama SPPG'] = df['Nama SPPG'].str.upper().str.strip()
df= df[df['Nama SPPG'].str.strip() != ""].reset_index(drop=True)

print(f"baris setelah cleansing:{len(df):,}")
```
