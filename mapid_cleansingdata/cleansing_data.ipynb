```python
import re

def cleaning_address(text):
    if pd.isna(text):
        return ""

    text = str(text) 
    text = re.sub(r'\b[A-Z0-9]{4}\+[A-Z0-9]{2,3}\b', '', text)  # hapus plus code 
    text = re.sub(r'[#@\^\$\?]', '', text)                        # hapus simbol 
    text = re.sub(r'^[,\s]+', '', text)                             # hapus awal koma 
    text = text.replace(' ,', ',')                                   # spasi sebelum koma 
    text = re.sub(r'\s+', ' ', text).strip() 
    return text 

df['Alamat SPPG'] = df['Alamat SPPG'].apply(cleaning_address)
df['Nama SPPG'] = df['Nama SPPG'].str.upper().str.strip()
df= df[df['Nama SPPG'].str.strip() != ""].reset_index(drop=True)

print(f"baris setelah cleansing:{len(df):,}")
```

```python
df['Latitude'] = pd.to_numeric(df['Latitude'], errors='coerce')
df['Longitude'] = pd.to_numeric(df['Longitude'], errors='coerce')

valid = (
    df['Latitude'].between(-11, 6) &
    df['Longitude'].between(95, 141.015074)
)

print(f"coordinate tidak valid: {(~valid).sum()}")
df['Koordinat Valid'] = valid
print(f"total data setelah validasi koordinat: {len(df):,}")

df_invalid = df[~valid]
df_invalid
```

```python
# Pastikan data adm sudah tersedia
adm = pd.read_csv('administrasi_kec_kab_kot.csv')
adm.columns = adm.columns.str.strip()

# Jika file adm dalam 1 kolom dengan pemisah ';', pecah menjadi 3 kolom
if adm.shape[1] == 1 and isinstance(adm.columns[0], str) and ';' in adm.columns[0]:
    adm = adm[adm.columns[0]].str.split(';', expand=True)
    adm.columns = ['Provinsi', 'Kab_Kota', 'Kecamatan']
else:
    adm.columns = [str(c).strip() for c in adm.columns]

# Ambil kolom yang dibutuhkan saja
adm = adm[['Provinsi', 'Kab_Kota', 'Kecamatan']].copy()
adm = adm.astype(str).apply(lambda col: col.str.strip())

# Normalisasi teks agar matching lebih konsisten
adm['Provinsi_norm'] = adm['Provinsi'].str.lower().str.replace(r'[^a-z0-9\s]', '', regex=True)
adm['Kab_Kota_norm'] = adm['Kab_Kota'].str.lower().str.replace(r'[^a-z0-9\s]', '', regex=True)
adm['Kecamatan_norm'] = adm['Kecamatan'].str.lower().str.replace(r'[^a-z0-9\s]', '', regex=True)

# Fungsi lookup wilayah berdasarkan alamat

def clean_text(value):
    if pd.isna(value):
        return ''
    return re.sub(r'[^a-z0-9\s]', '', str(value).lower())


def wilayah(alamat):
    if pd.isna(alamat):
        return None, None, None

    address = clean_text(alamat)
    if address == '':
        return None, None, None

    matches = adm[adm['Kecamatan_norm'].apply(lambda kec: kec in address)].copy()
    if matches.empty:
        return None, None, None

    best = matches.loc[matches['Kecamatan_norm'].str.len().idxmax()]
    return best['Provinsi'], best['Kab_Kota'], best['Kecamatan']

# Pastikan kolom target ada
for col in ['Provinsi SPPG', 'Kabupaten/Kota SPPG', 'Kecamatan SPPG']:
    if col not in df.columns:
        df[col] = None

# Isi data wilayah dari adm berdasarkan alamat
loc = df['Alamat SPPG'].apply(
    lambda x: pd.Series(
        wilayah(x),
        index=['Provinsi SPPG', 'Kabupaten/Kota SPPG', 'Kecamatan SPPG']
    )
)

df[['Provinsi SPPG', 'Kabupaten/Kota SPPG', 'Kecamatan SPPG']] = loc

berhasil = df['Kecamatan SPPG'].notna().sum()
kosong = df['Kecamatan SPPG'].isna().sum()

print(f"join berhasil : {berhasil} ({(berhasil / len(df) * 100):.0f}%)")
print(f"perlu isi manual : {kosong}")

if kosong > 0:
    print(df[df['Kecamatan SPPG'].isna()][['Nama SPPG', 'Alamat SPPG', 'link']].to_string(index=False))
```

```python
%pip install rapidfuzz

# Fuzzy matching untuk sisa data yang belum terhubung
# Tujuan: menangani typo, singkatan, dan variasi penulisan nama kecamatan

try:
    from rapidfuzz import process, fuzz
    HAS_RAPIDFUZZ = True
except Exception:
    HAS_RAPIDFUZZ = False
    from difflib import SequenceMatcher

# Pastikan tabel master wilayah memiliki kolom normalisasi
if 'Kecamatan_norm' not in adm.columns:
    adm['Kecamatan_norm'] = (
        adm['Kecamatan'].fillna('').astype(str)
        .str.lower()
        .str.replace(r'[^a-z0-9\s]', ' ', regex=True)
        .str.replace(r'\s+', ' ', regex=True)
        .str.strip()
    )


def normalize_wilayah(text):
    if pd.isna(text):
        return ''
    s = str(text).lower()
    s = s.replace('kec.', 'kecamatan')
    s = s.replace('kab.', 'kabupaten')
    s = s.replace('kota ', 'kota ')
    s = re.sub(r'[^a-z0-9\s]', ' ', s)
    s = re.sub(r'\s+', ' ', s).strip()
    return s


def fuzzy_match_kecamatan(alamat):
    if pd.isna(alamat):
        return (None, None, None)

    address = normalize_wilayah(alamat)
    if address == '':
        return (None, None, None)

    # 1. Cek kecamatan yang muncul langsung di alamat
    matches = adm[adm['Kecamatan_norm'].apply(lambda x: x in address)].copy()
    if not matches.empty:
        best = matches.loc[matches['Kecamatan_norm'].str.len().idxmax()]
        return (best['Provinsi'], best['Kab_Kota'], best['Kecamatan'])

    # 2. Fuzzy matching untuk typo/singkatan/variasi ejaan
    if HAS_RAPIDFUZZ:
        candidates = process.extract(
            address,
            adm['Kecamatan_norm'].tolist(),
            scorer=fuzz.token_set_ratio,
            limit=10,
        )
        if candidates and candidates[0][1] >= 75:
            best_name = candidates[0][0]
            best = adm.loc[adm['Kecamatan_norm'] == best_name, ['Provinsi', 'Kab_Kota', 'Kecamatan']].iloc[0]
            return (best['Provinsi'], best['Kab_Kota'], best['Kecamatan'])
    else:
        best_score = 0
        best_row = None
        for _, row in adm.iterrows():
            score = SequenceMatcher(None, address, row['Kecamatan_norm']).ratio()
            if score > best_score:
                best_score = score
                best_row = row
        if best_row is not None and best_score >= 0.75:
            return (best_row['Provinsi'], best_row['Kab_Kota'], best_row['Kecamatan'])

    return (None, None, None)

# Hanya proses data yang masih kosong
mask = df['Kecamatan SPPG'].fillna('').astype(str).str.strip().eq('')

if mask.any():
    hasil = pd.DataFrame(
        df.loc[mask, 'Alamat SPPG'].apply(fuzzy_match_kecamatan).tolist(),
        index=df.index[mask],
        columns=['Provinsi SPPG', 'Kabupaten/Kota SPPG', 'Kecamatan SPPG'],
    )

    df.loc[mask, ['Provinsi SPPG', 'Kabupaten/Kota SPPG', 'Kecamatan SPPG']] = hasil

    match_sebelum = int(df['Kecamatan SPPG'].notna().sum())
    # Hitung sisa data yang belum terisi setelah fuzzy
    sisa = df[df['Kecamatan SPPG'].fillna('').astype(str).str.strip().eq('')]

    print(f'jumlah match yang sudah ada: {match_sebelum}')
    print(f'sisa data manual setelah fuzzy: {len(sisa)}')
    display(sisa[['Nama SPPG', 'Alamat SPPG', 'link']].head(20))
else:
    print('tidak ada data yang perlu diproses fuzzy matching')

# Catatan:
# - Jika hasil fuzzy masih belum sempurna, lanjutkan dengan pengisian manual untuk data yang tersisa.
# - Ambang skor default 75 bisa disesuaikan sesuai kualitas data: 70-80 untuk data lebih longgar.
```
