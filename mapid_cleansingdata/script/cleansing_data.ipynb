```python
import re

def cleaning_address(text):
    if pd.isna(text):
        return ""

    text = str(text) 
    text = re.sub(r'\b[A-Z0-9]{4}\+[A-Z0-9]{2,3}\b', '', text)  # hapus plus code 
    text = re.sub(r'[#@\^\$\?]', '', text)                        # hapus simbol 
    text = re.sub(r'^[,\s]+', '', text)                             # hapus awal koma 
    text = text.replace(' ,', ',')                                   # spasi sebelum koma 
    text = re.sub(r'\s+', ' ', text).strip() 
    return text 

df['Alamat SPPG'] = df['Alamat SPPG'].apply(cleaning_address)
df['Nama SPPG'] = df['Nama SPPG'].str.upper().str.strip()
df= df[df['Nama SPPG'].str.strip() != ""].reset_index(drop=True)

print(f"baris setelah cleansing:{len(df):,}")
```

```python
import re

# Koreksi nama SPPG dan wilayah berdasarkan sumber Google Maps yang paling valid
# agar nama SPPG, kecamatan, kabupaten/kota, dan provinsi sesuai dengan alamat asli di Maps

def parse_address_location(address):
    if pd.isna(address):
        return {"Provinsi SPPG": None, "Kabupaten/Kota SPPG": None, "Kecamatan SPPG": None}

    s = str(address)
    s = s.replace('Kec.', 'Kecamatan ')
    s = s.replace('Kecamaten', 'Kecamatan')
    s = s.replace('Kab.', 'Kabupaten ')
    s = s.replace('Kabupaten ', 'Kabupaten ')

    kec = None
    for pat in [r'Kecamatan\s+([^,]+)', r'Kec\.?\s*([^,]+)']:
        m = re.search(pat, s, flags=re.I)
        if m:
            kec = m.group(1).strip()
            break

    kab = None
    for pat in [r'Kabupaten\s+([^,]+)', r'Kota\s+([^,]+)']:
        m = re.search(pat, s, flags=re.I)
        if m:
            kab = m.group(1).strip()
            break

    prov = None
    for token in [
        'Aceh', 'Sumatera Utara', 'Sumatera Barat', 'Riau', 'Jambi', 'Sumatera Selatan',
        'Bengkulu', 'Lampung', 'Bangka Belitung', 'Kepulauan Riau', 'Jakarta', 'Jawa Barat',
        'Jawa Tengah', 'Jawa Timur', 'DIY', 'Yogyakarta', 'Bali', 'NTT', 'NTB',
        'Kalimantan Barat', 'Kalimantan Tengah', 'Kalimantan Selatan', 'Kalimantan Timur',
        'Kalimantan Utara', 'Sulawesi Utara', 'Sulawesi Tengah', 'Sulawesi Selatan',
        'Sulawesi Tenggara', 'Gorontalo', 'Maluku', 'Maluku Utara', 'Papua', 'Papua Barat'
    ]:
        if re.search(rf'{re.escape(token)}', s, flags=re.I):
            prov = token
            break

    return {
        'Provinsi SPPG': prov,
        'Kabupaten/Kota SPPG': kab,
        'Kecamatan SPPG': kec,
    }

raw_google = pd.read_csv(input_file, encoding='utf-8-sig', low_memory=False)
raw_google = raw_google.dropna(subset=['title']).reset_index(drop=True)
raw_google = raw_google.iloc[:len(df)].reset_index(drop=True)

# pakai title/address dari raw Google Maps sebagai sumber kebenaran
# agar nama SPPG dan wilayah sesuai dengan lokasi di Maps

df['Nama SPPG'] = raw_google['title'].astype(str).str.strip()
df['Alamat SPPG'] = raw_google['address'].fillna('').astype(str)
df['Latitude'] = pd.to_numeric(raw_google['latitude'], errors='coerce')
df['Longitude'] = pd.to_numeric(raw_google['longitude'], errors='coerce')

parsed = raw_google['address'].apply(parse_address_location)
parsed_df = pd.DataFrame(parsed.tolist())
df[['Provinsi SPPG', 'Kabupaten/Kota SPPG', 'Kecamatan SPPG']] = parsed_df[['Provinsi SPPG', 'Kabupaten/Kota SPPG', 'Kecamatan SPPG']]
df['Koordinat Valid'] = df['Latitude'].between(-11, 6) & df['Longitude'].between(95, 141.015074)

df[['Nama SPPG', 'Provinsi SPPG', 'Kabupaten/Kota SPPG', 'Kecamatan SPPG', 'Latitude', 'Longitude']].head()
```

```python
# Pastikan data adm sudah tersedia
adm = pd.read_csv('../data/administrasi_kec_kab_kot.csv')
adm.columns = adm.columns.str.strip()

# Jika file adm dalam 1 kolom dengan pemisah ';', pecah menjadi 3 kolom
if adm.shape[1] == 1 and isinstance(adm.columns[0], str) and ';' in adm.columns[0]:
    adm = adm[adm.columns[0]].str.split(';', expand=True)
    adm.columns = ['Provinsi', 'Kab_Kota', 'Kecamatan']
else:
    adm.columns = [str(c).strip() for c in adm.columns]

# Ambil kolom yang dibutuhkan saja
adm = adm[['Provinsi', 'Kab_Kota', 'Kecamatan']].copy()
adm = adm.astype(str).apply(lambda col: col.str.strip())

# Normalisasi teks agar matching lebih konsisten
adm['Provinsi_norm'] = adm['Provinsi'].str.lower().str.replace(r'[^a-z0-9\s]', '', regex=True)
adm['Kab_Kota_norm'] = adm['Kab_Kota'].str.lower().str.replace(r'[^a-z0-9\s]', '', regex=True)
adm['Kecamatan_norm'] = adm['Kecamatan'].str.lower().str.replace(r'[^a-z0-9\s]', '', regex=True)

# Fungsi lookup wilayah berdasarkan alamat

def clean_text(value):
    if pd.isna(value):
        return ''
    return re.sub(r'[^a-z0-9\s]', '', str(value).lower())


def wilayah(alamat):
    if pd.isna(alamat):
        return None, None, None

    address = clean_text(alamat)
    if address == '':
        return None, None, None

    matches = adm[adm['Kecamatan_norm'].apply(lambda kec: kec in address)].copy()
    if matches.empty:
        return None, None, None

    best = matches.loc[matches['Kecamatan_norm'].str.len().idxmax()]
    return best['Provinsi'], best['Kab_Kota'], best['Kecamatan']

# Pastikan kolom target ada
for col in ['Provinsi SPPG', 'Kabupaten/Kota SPPG', 'Kecamatan SPPG']:
    if col not in df.columns:
        df[col] = None

# Isi data wilayah dari adm berdasarkan alamat
loc = df['Alamat SPPG'].apply(
    lambda x: pd.Series(
        wilayah(x),
        index=['Provinsi SPPG', 'Kabupaten/Kota SPPG', 'Kecamatan SPPG']
    )
)

df[['Provinsi SPPG', 'Kabupaten/Kota SPPG', 'Kecamatan SPPG']] = loc

berhasil = df['Kecamatan SPPG'].notna().sum()
kosong = df['Kecamatan SPPG'].isna().sum()

print(f"join berhasil : {berhasil} ({(berhasil / len(df) * 100):.0f}%)")
print(f"perlu isi manual : {kosong}")

if kosong > 0:
    print(df[df['Kecamatan SPPG'].isna()][['Nama SPPG', 'Alamat SPPG', 'link']].to_string(index=False))
```

```python
%pip install rapidfuzz

# Fuzzy matching untuk sisa data yang belum terhubung
# Tujuan: menangani typo, singkatan, dan variasi penulisan nama kecamatan

try:
    from rapidfuzz import process, fuzz
    HAS_RAPIDFUZZ = True
except Exception:
    HAS_RAPIDFUZZ = False
    from difflib import SequenceMatcher

# Pastikan tabel master wilayah memiliki kolom normalisasi
if 'Kecamatan_norm' not in adm.columns:
    adm['Kecamatan_norm'] = (
        adm['Kecamatan'].fillna('').astype(str)
        .str.lower()
        .str.replace(r'[^a-z0-9\s]', ' ', regex=True)
        .str.replace(r'\s+', ' ', regex=True)
        .str.strip()
    )


def normalize_wilayah(text):
    if pd.isna(text):
        return ''
    s = str(text).lower()
    s = s.replace('kec.', 'kecamatan')
    s = s.replace('kab.', 'kabupaten')
    s = s.replace('kota ', 'kota ')
    s = re.sub(r'[^a-z0-9\s]', ' ', s)
    s = re.sub(r'\s+', ' ', s).strip()
    return s


def fuzzy_match_kecamatan(alamat):
    if pd.isna(alamat):
        return (None, None, None)

    address = normalize_wilayah(alamat)
    if address == '':
        return (None, None, None)

    # 1. Cek kecamatan yang muncul langsung di alamat
    matches = adm[adm['Kecamatan_norm'].apply(lambda x: x in address)].copy()
    if not matches.empty:
        best = matches.loc[matches['Kecamatan_norm'].str.len().idxmax()]
        return (best['Provinsi'], best['Kab_Kota'], best['Kecamatan'])

    # 2. Fuzzy matching untuk typo/singkatan/variasi ejaan
    if HAS_RAPIDFUZZ:
        candidates = process.extract(
            address,
            adm['Kecamatan_norm'].tolist(),
            scorer=fuzz.token_set_ratio,
            limit=10,
        )
        if candidates and candidates[0][1] >= 75:
            best_name = candidates[0][0]
            best = adm.loc[adm['Kecamatan_norm'] == best_name, ['Provinsi', 'Kab_Kota', 'Kecamatan']].iloc[0]
            return (best['Provinsi'], best['Kab_Kota'], best['Kecamatan'])
    else:
        best_score = 0
        best_row = None
        for _, row in adm.iterrows():
            score = SequenceMatcher(None, address, row['Kecamatan_norm']).ratio()
            if score > best_score:
                best_score = score
                best_row = row
        if best_row is not None and best_score >= 0.75:
            return (best_row['Provinsi'], best_row['Kab_Kota'], best_row['Kecamatan'])

    return (None, None, None)

# Hanya proses data yang masih kosong
mask = df['Kecamatan SPPG'].fillna('').astype(str).str.strip().eq('')

if mask.any():
    hasil = pd.DataFrame(
        df.loc[mask, 'Alamat SPPG'].apply(fuzzy_match_kecamatan).tolist(),
        index=df.index[mask],
        columns=['Provinsi SPPG', 'Kabupaten/Kota SPPG', 'Kecamatan SPPG'],
    )

    df.loc[mask, ['Provinsi SPPG', 'Kabupaten/Kota SPPG', 'Kecamatan SPPG']] = hasil

    match_sebelum = int(df['Kecamatan SPPG'].notna().sum())
    # Hitung sisa data yang belum terisi setelah fuzzy
    sisa = df[df['Kecamatan SPPG'].fillna('').astype(str).str.strip().eq('')]

    print(f'jumlah match yang sudah ada: {match_sebelum}')
    print(f'sisa data manual setelah fuzzy: {len(sisa)}')
    display(sisa[['Nama SPPG', 'Alamat SPPG', 'link']].head(20))
else:
    print('tidak ada data yang perlu diproses fuzzy matching')

# Catatan:
# - Jika hasil fuzzy masih belum sempurna, lanjutkan dengan pengisian manual untuk data yang tersisa.
# - Ambang skor default 75 bisa disesuaikan sesuai kualitas data: 70-80 untuk data lebih longgar.
```

```python
# Filter baris berdasarkan nama SPPG yang valid
# Tidak mengubah kolom alamat, provinsi, kabupaten/kota, kecamatan, latitude, longitude

# Normalisasi nama agar pencocokan lebih konsisten
df["Nama SPPG_norm"] = (
    df["Nama SPPG"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.replace("satuam", "satuan", regex=False)
    .str.replace(r"[^a-z0-9\s]", " ", regex=True)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

# Keyword yang dianggap valid untuk SPPG
mask = (
    df["Nama SPPG_norm"].str.contains(r"\bsppg\b", regex=True) |
    df["Nama SPPG_norm"].str.contains(r"satuan pelayanan pemenuhan gizi", regex=False) |
    df["Nama SPPG_norm"].str.contains(r"dapur", regex=False) |
    df["Nama SPPG_norm"].str.contains(r"mbg", regex=False) |
    df["Nama SPPG_norm"].str.contains(r"dapur sppg", regex=False)
)

filtered_df = df.loc[mask].copy()

# Pertahankan data asli pada kolom selain Nama SPPG
filtered_df = filtered_df.drop(columns=["Nama SPPG_norm"], errors="ignore").reset_index(drop=True)

print(f"jumlah sebelum filter: {len(df):,}")
print(f"jumlah setelah filter Nama SPPG: {len(filtered_df):,}")
print(f"persentase: {(len(filtered_df) / max(len(df), 1) * 100):.2f}%")

# Cek apakah jumlahnya sudah mendekati target
if len(filtered_df) >= 6200 and len(filtered_df) <= 6300:
    print("Jumlah hasil filter sesuai estimasi target (~6243).")
else:
    print("Jumlah hasil filter tidak sesuai estimasi target, perlu pengecekan keyword/manual review.")

# Tampilkan beberapa contoh
display(filtered_df.head(10))

# Simpan hasil filter ke file baru
filtered_output = "filtered_nama_sppg_cleanser1.csv"
filtered_df.to_csv(filtered_output, index=False, encoding="utf-8-sig")
print(f"File filter tersimpan: {filtered_output}")
```
