```python
import os
import pandas as pd

# 1. Load file sumber
candidates = [
    os.path.join(os.getcwd(), 'sort_of_sppg', 'master_cleaned_sppg_final_10175.csv'),
    os.path.join(os.getcwd(), 'master_cleaned_sppg_final_10175.csv'),
    r'C:\kerjaan_magang\sort_of_sppg\master_cleaned_sppg_final_10175.csv',
]
src = next((p for p in candidates if os.path.exists(p)), candidates[0])
out_dir = os.path.dirname(src)
out_file = os.path.join(out_dir, 'master_cleaned_sppg_prefinal_sorted.csv')

keywords = ['sppg', 'dapur', 'satuan pelayanan pemenuhan gizi']


def has_keyword(row):
    text = ' '.join(str(v) for v in row.fillna('').astype(str)).lower()
    return any(k in text for k in keywords)


df = pd.read_csv(src, engine='python', on_bad_lines='skip')
mask = df.apply(has_keyword, axis=1)
filtered = df[mask].copy()
remaining = df[~mask].copy()

try:
    filtered.to_csv(out_file, index=False)
    print(f'Jumlah data total: {len(df)}')
    print(f'Jumlah data dengan keyword SPPG/Dapur/Satuan Pelayanan Pemenuhan Gizi: {len(filtered)}')
    print(f'Jumlah data tanpa keyword: {len(remaining)}')
    print(f'File output disimpan: {out_file}')
except PermissionError:
    fallback = os.path.join(out_dir, 'master_cleaned_sppg_prefinal_sorted_safe.csv')
    filtered.to_csv(fallback, index=False)
    print(f'Permission denied untuk {out_file}. File disimpan ke fallback: {fallback}')
    print(f'Jumlah data total: {len(df)}')
    print(f'Jumlah data dengan keyword SPPG/Dapur/Satuan Pelayanan Pemenuhan Gizi: {len(filtered)}')
    print(f'Jumlah data tanpa keyword: {len(remaining)}')
```

```python
import os
import pandas as pd

fp = os.path.join(os.getcwd(), 'sort_of_sppg', 'master_cleaned_sppg_prefinal_sorted.csv')

if os.path.exists(fp):
    df = pd.read_csv(fp, engine='python', on_bad_lines='skip')
    for col in df.columns:
        df[col] = df[col].map(lambda x: x.replace('"', '').replace("'", '') if isinstance(x, str) else x)

    try:
        df.to_csv(fp, index=False)
        print(f'File diperbarui: {fp}')
    except PermissionError:
        fallback = fp.replace('.csv', '_cleaned.csv')
        df.to_csv(fallback, index=False)
        print(f'Permission denied untuk {fp}. File dibersihkan ke: {fallback}')

    print(f'Jumlah baris setelah pembersihan: {len(df)}')
    print(f'Jumlah kolom: {len(df.columns)}')
else:
    print(f'File tidak ditemukan: {fp}')
```

```python
import os
import pandas as pd

candidates = [
    os.path.join(os.getcwd(), 'sort_of_sppg', 'master_cleaned_sppg_final_10175.csv'),
    os.path.join(os.getcwd(), 'master_cleaned_sppg_final_10175.csv'),
    r'C:\kerjaan_magang\sort_of_sppg\master_cleaned_sppg_final_10175.csv',
]

src = next((p for p in candidates if os.path.exists(p)), candidates[0])
keywords = ['sppg', 'dapur', 'satuan pelayanan pemenuhan gizi']


def has_keyword(row):
    text = ' '.join(str(v) for v in row.fillna('').astype(str)).lower()
    return any(k in text for k in keywords)


df = pd.read_csv(src, engine='python', on_bad_lines='skip')
mask = df.apply(has_keyword, axis=1)
non_keyword_rows = df[~mask].copy().reset_index(drop=True)

out_file = os.path.join(os.path.dirname(src), 'master_cleaned_sppg_non_keyword.csv')
non_keyword_rows.to_csv(out_file, index=False)
print(f'Jumlah data total: {len(df)}')
print(f'Jumlah data tanpa keyword: {len(non_keyword_rows)}')
print(f'File output tanpa keyword: {out_file}')
```
